```python
##Q7: Sum of weights
```

```python
df_asia = df[df['origin'] == 'Asia']
df_asia
```

```python
df_asia = df_asia[['vehicle_weight', 'model_year']]
df_asia
```

```python
df_asia.head(7)
```

```python
x = df_asia.head(7).values
x
```

```python
xT = x.T
xT
```

```python
XTX = xT.dot(x)
XTX
```

```python
import numpy as np
```

```python
XTXi = np.linalg.inv(XTX)
XTXi
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = XTXi.dot(xT).dot(y)
w
```

```python
result = w[0] + w[1]
result
```
