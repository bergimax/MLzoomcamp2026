```python
categorical_variables = ['make', 'engine_fuel_type', 'transmission_type', 'driven_wheels',
       'market_category', 'vehicle_size', 'vehicle_style'
]
```

```python
categories = {}

for c in categorical_variables:
    categories[c] = list(df[c].value_counts().head().index)

categories
```

```python
def prepare_X(df):
    df= df.copy()
    features = base.copy()
    
    df['age'] = 2017 - df.year
    features.append('age')
    
    for i in [2, 3, 4]:
        df['num_door_%s' %i] = (df.number_of_doors== i).astype('int')
        features.append('num_door_%s' %i)

    for name, values in categories.items():
        for val in values:
            df['%s_%s' % (name, val)] = (df[name] == val).astype(int)
            features.append('%s_%s' % (name,val))
            
    df_num = df[features]
    df_num = df_num.fillna(0)
    X = df_num.values
    return X
```

```python
#train
X_train = prepare_X(df_train)
w0, w = train_linear_regression(X_train, y_train)
#validaton
X_val = prepare_X(df_val)
y_pred= w0 + X_val.dot(w)
#evaluatio
rmse(y_val, y_pred)
```

```python
##Regularization
```

```python
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ ones, X])
    XTX = X.T.dot(X)
    XTX = XTX + r*np.eye(XTX.shape[0])
    XTX_i= np.linalg.inv(XTX)

    w_full = XTX_i.dot(X.T).dot(y)
    return w_full[0], w_full[1:]
```

```python
#train
X_train = prepare_X(df_train)
w0, w = train_linear_regression_reg(X_train, y_train, r=0.01)
#validaton
X_val = prepare_X(df_val)
y_pred= w0 + X_val.dot(w)
#evaluatio
rmse(y_val, y_pred)
```
